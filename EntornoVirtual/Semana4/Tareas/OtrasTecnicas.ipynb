```python
import pandas as pd


url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

print(df.head())
```

**Asimetría (Skewness)**

La función df["Age"].skew() mide si la distribución de edades está sesgada hacia la derecha o la izquierda. Si el valor es positivo, hay más pasajeros jóvenes con algunos mayores extremos; si es negativo, la cola está hacia los más jóvenes. En general, esto nos ayuda a ver si la población del barco era más joven o más adulta.

```python
df["Age"].skew()
```

**Curtosis (Kurtosis)**

La función df["Fare"].kurt() indica cuán “pesadas” son las colas de la distribución del precio del pasaje. Un valor alto sugiere que existen precios muy extremos, es decir, outliers. Esto confirma que hay algunos pasajes mucho más caros que el resto.

```python
df["Fare"].kurt()
```

```python
pd.crosstab(df["Pclass"], df["Survived"])
```

**Tablas Cruzadas (Cross-Tabulation)**

pd.crosstab(df["Pclass"], df["Survived"]) permite comparar la clase del pasajero con su supervivencia. Al cruzar ambas variables, se observa rápidamente que los pasajeros de primera clase sobrevivieron más, mientras que los de tercera clase tuvieron menor tasa de supervivencia. Esto muestra una relación clara entre 

```python
pd.crosstab(df["Pclass"], df["Survived"], normalize="index")
```
