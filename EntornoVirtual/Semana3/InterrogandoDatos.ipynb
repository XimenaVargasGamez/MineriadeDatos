```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

```python
# A:
survival_rate = df['Survived'].value_counts(normalize=True)
print(survival_rate.to_string())
print(f"\ntasa de supervivencia: {survival_rate[1]*100:.2f}%")
print(f"tasa de mortalidad: {survival_rate[0]*100:.2f}%")
```

```python
# B:
gender_survival = df.groupby('Sex')['Survived'].mean()
print(gender_survival.to_string())
print(f"mujeres sobrevivientes: {gender_survival['female']*100:.2f}%")
print(f"hombres sobrevivientes: {gender_survival['male']*100:.2f}%")
```

```python
# C:
q1, q3 = df['Fare'].quantile([0.25, 0.75])
iqr = q3 - q1
limite_superior = q3 + 1.5 * iqr
print(f"Q1: {q1}, Q3: {q3}")
print(f"IQR: {iqr}")
print(f"lim superior: {limite_superior}")
outliers = df[df['Fare'] > limite_superior]
print(f"pasajeros con tarifa > límite: {len(outliers)}")
print(f"outliers:\n{outliers['Pclass'].value_counts().to_string()}")
```

```python
# D:
media_fare = df['Fare'].mean()
mediana_fare = df['Fare'].median()
print(f"media: {media_fare:.2f}, mediana: {mediana_fare:.2f}")
print("la  distribución no es normal")
```

```python
# E:
muestra_estratificada = df.groupby('Survived', group_keys=False).apply(
    lambda x: x.sample(n=int(75 * len(x) / len(df)), random_state=42)
)
print(f"tamano de muestra: {len(muestra_estratificada)}")
print(muestra_estratificada['Survived'].value_counts(normalize=True).to_string())
```

```python

# F:
edad_por_supervivencia = df.groupby('Survived')['Age'].mean()
print(f"edad promedio por supervivencia:\n{edad_por_supervivencia.to_string()}")
```

```python
# H:
df['Title'] = df['Name'].str.extract(r'([A-Za-z]+)\.', expand=False)
edad_por_titulo = df.groupby('Title')['Age'].median()
print(f"edad mediana por titulo:\n{edad_por_titulo.to_string()}")
```

```python
# I:
varianza_survived = df['Survived'].var()
print(f"varianza de sobrevivientes: {varianza_survived:.4f}")
```

```python
# J:
agrupacion_triple = df.groupby(['Pclass', 'Sex', 'Embarked'])['PassengerId'].count()
print(agrupacion_triple.to_string())
print(f"subgrupos con 1 pasajero: {(agrupacion_triple == 1).sum()}")
```
