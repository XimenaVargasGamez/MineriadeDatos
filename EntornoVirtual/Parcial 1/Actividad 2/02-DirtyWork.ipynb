```python
import random

random_list = [random.random() for _ in range(5_000_000)]
print("random list length:", len(random_list))
```

```python
def mean_custom(values):
    accumulate = 0
    for val in values:
        accumulate += val
    return accumulate / len(values)

def st_deviation(values):
    sd=0
    mean_val = mean_custom(values)
    for val in values:
        sd += (val - mean_val) ** 2

    sd = sd / len(values)

    return sd ** 0.5

print(f'Std: {st_deviation(random_list)}')
```

```python
import numpy as np

np_array = np.array(random_list, dtype=np.float64)

print(f'Std: {np_array.std()}')
```
